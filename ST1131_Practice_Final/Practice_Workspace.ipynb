## Question 11 (4 marks)
**Data-cleaning pipeline. One mark per answer.** Read the car-data instructions on the section cover. Complete the code so it performs the specified cleaning exactly. A and B must be valid Python expressions, including quotes where needed.

(a) `__A__` = __________
(b) `__B__` = __________
(c) Number of rows removed by the duplicate-Record_ID step only: __________
(d) Final number of rows in `cars_clean`: __________

```python
raw = pd.read_csv("data/cars_exam_raw.csv")
cars = raw.drop_duplicates(subset=__A__, keep="first").copy()
numeric_cols = ["Selling_Price", "Present_Price", "Kms_Driven"]
for col in numeric_cols:
    cars[col] = pd.to_numeric(cars[col], errors=__B__)
cars["Transmission"] = cars["Transmission"].str.strip().str.lower()
cars["Fuel_Type"] = cars["Fuel_Type"].str.strip().str.upper()
required = numeric_cols + ["Transmission", "Fuel_Type"]
cars = cars.dropna(subset=required)
cars = cars[(cars["Selling_Price"] > 0) &
            (cars["Present_Price"] > 0) &
            (cars["Kms_Driven"] >= 0)]
cars = cars[cars["Transmission"].isin(["automatic", "manual"]) &
            cars["Fuel_Type"].isin(["CNG", "DIESEL", "PETROL"])]
cars_clean = cars.reset_index(drop=True)
```
