Some important functions we will use are:

- `stats.norm.ppf(q)` gives the $q$-th quantile of $N(0, 1)$, which we use as the multiplier $q_{1 - \alpha/2}$ for a confidence interval for a proportion.
- `stats.t.ppf(q, df)` gives the $q$-th quantile of the $t$-distribution with `df` degrees of freedom, which we use as the multiplier $t_{n-1,\,1-\alpha/2}$ for a confidence interval for a mean.
- `stats.t.interval(level, df, loc, scale)` returns a confidence interval for a mean directly, where `loc` is $\bar{X}$ and `scale` is the standard error $s/\sqrt{n}$.
