## B. `iloc` = positions

Think **integer location**.

Use when the question says:
- first row
- rows 2 to 5
- column in position 3


````python
# First row
df.iloc[0]

# First 3 rows
df.iloc[0:3]

# First 3 rows, first 2 columns
df.iloc[0:3, 0:2]

# One specific cell: first row, second column
df.iloc[0, 1]
````

### Important `iloc` slicing rule

`0:3` means positions **0, 1, 2**.  
The ending position `3` is **not included**.


## C. `loc` = labels

Use names / labels rather than positions.


````python
# All rows, selected columns
df.loc[:, ["gender", "height"]]

# Rows with labels 0, 1, 2 and selected columns
df.loc[0:2, ["gender", "height"]]

# One specific cell
df.loc[0, "height"]
````

### `loc` vs `iloc`

| Need | Use |
|---|---|
| position number | `iloc` |
| row/column label | `loc` |

Easy memory trick: **i**loc = **i**nteger.


# 3. Filtering rows with conditions

This is one of the most common pandas exam patterns.

## A. One condition


````python
# Keep only rows where height > 165
df[df["height"] > 165]
````

## B. Two conditions: AND

Use `&`, not Python's `and`.


````python
df[(df["height"] > 165) & (df["score"] > 80)]
````

## C. Two conditions: OR

Use `|`, not Python's `or`.


````python
df[(df["gender"] == "F") | (df["score"] > 90)]
````

## D. NOT

Use `~`.


````python
df[~(df["smoker"] == "Yes")]
````

## E. Membership: `.isin()`

Use when you want rows matching one of several categories.


````python
df[df["group"].isin(["A", "B"])]
````

## F. Between two values

Both approaches are useful.


````python
# Method 1
df[(df["score"] >= 75) & (df["score"] <= 90)]

# Method 2
df[df["score"].between(75, 90)]
````

### Filtering trap

Correct:
```python
df[(df["height"] > 165) & (df["score"] > 80)]
```

Wrong:
```python
df[df["height"] > 165 and df["score"] > 80]
```

For pandas Series, use:
- `&` = AND
- `|` = OR
- `~` = NOT

and put **each condition inside brackets**.

# 6. NumPy essentials

Use NumPy when you are working directly with arrays or need numerical functions.


````python
x = np.array([10, 20, 30, 40, 50])

# Basic summaries
np.sum(x)
np.mean(x)
np.median(x)
np.min(x)
np.max(x)

# Sample standard deviation and variance
np.std(x, ddof=1)
np.var(x, ddof=1)

# Quartiles / quantiles
np.quantile(x, 0.25)   # Q1
np.quantile(x, 0.50)   # median
np.quantile(x, 0.75)   # Q3
````

### Why `ddof=1`?

For a **sample variance / sample standard deviation**, ST1131 uses denominator `n - 1`.

So:
```python
np.std(x, ddof=1)
np.var(x, ddof=1)
```

Pandas `.std()` and `.var()` already use the sample version by default.


## NumPy boolean filtering


````python
x = np.array([12, 18, 25, 31, 40])

# Values greater than 20
x[x > 20]

# Values between 18 and 31
x[(x >= 18) & (x <= 31)]

# Count satisfying a condition
np.sum(x > 20)

# Proportion satisfying a condition
np.mean(x > 20)
````

# 8. Numerical summaries

Use for **quantitative variables**.

## Individual commands


````python
df["height"].mean()
df["height"].median()
df["height"].std()
df["height"].var()
df["height"].min()
df["height"].max()

# Range
df["height"].max() - df["height"].min()

# Quartiles
df["height"].quantile(0.25)
df["height"].quantile(0.50)
df["height"].quantile(0.75)

# IQR
q1 = df["height"].quantile(0.25)
q3 = df["height"].quantile(0.75)
iqr = q3 - q1
iqr
````

## Several summaries at once with `agg()`


````python
df["height"].agg([
    "count",
    "mean",
    "median",
    "std",
    "var",
    "min",
    "max"
])
````

### When to prefer mean/SD vs median/IQR

- roughly symmetric / bell-shaped → **mean + standard deviation**
- skewed / contains outliers → **median + IQR**

`alpha` ranges from 0 to 1:
- `1` = fully opaque
- smaller values = more transparent

# 12. TWO QUANTITATIVE VARIABLES → SCATTER PLOT

Use when **both variables are numerical**.

Usually:
- explanatory variable → x-axis
- response variable → y-axis


````python
plt.scatter(df["height"], df["weight"])

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Weight vs Height")

plt.show()
````

## pandas version


````python
df.plot(
    kind="scatter",
    x="height",
    y="weight"
)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Weight vs Height")
plt.show()
````

## Scatter plot with grouped categories + legend

Use when you have:
- x = quantitative
- y = quantitative
- another categorical variable to distinguish groups


````python
for gender, subdf in df.groupby("gender"):
    plt.scatter(
        subdf["height"],
        subdf["weight"],
        label=gender
    )

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.title("Weight vs Height by Gender")

# Shows which group each point belongs to
plt.legend(title="Gender")

plt.show()
````

## Scatter plot interpretation checklist

Check:
1. association present?
2. positive or negative?
3. approximately linear?
4. strong / moderate / weak?
5. unusual observations / outliers?
6. does variability stay roughly constant or fan out?

# 14. TWO CATEGORICAL VARIABLES → CONTINGENCY TABLE

Use `pd.crosstab()`.

Example:
- gender
- smoker


````python
table = pd.crosstab(df["gender"], df["smoker"])
table
````

## Row-wise percentages

Each **row sums to 100%**.

Use when rows are the explanatory groups and you want to compare the response distribution **within each row group**.


````python
row_pct = pd.crosstab(
    df["gender"],
    df["smoker"],
    normalize="index"
) * 100

row_pct
````

## Column-wise percentages

Each **column sums to 100%**.


````python
col_pct = pd.crosstab(
    df["gender"],
    df["smoker"],
    normalize="columns"
) * 100

col_pct
````

## Overall proportions

Entire table sums to 1.


````python
pd.crosstab(
    df["gender"],
    df["smoker"],
    normalize=True
)
````

# 15. TWO CATEGORICAL VARIABLES → GROUPED / STACKED BAR PLOTS

## Grouped / clustered bar plot


````python
table = pd.crosstab(df["gender"], df["smoker"])

table.plot(kind="bar")

plt.xlabel("Gender")
plt.ylabel("Frequency")
plt.title("Smoking Status by Gender")
plt.xticks(rotation=0)
plt.legend(title="Smoking status")

plt.show()
````

## 100% stacked bar plot

Use percentages, not raw counts, when comparing proportions between groups.


````python
row_pct = pd.crosstab(
    df["gender"],
    df["smoker"],
    normalize="index"
) * 100

row_pct.plot(kind="bar", stacked=True)

plt.xlabel("Gender")
plt.ylabel("Percentage")
plt.title("Smoking Status by Gender")
plt.xticks(rotation=0)
plt.legend(title="Smoking status")

plt.show()
````
