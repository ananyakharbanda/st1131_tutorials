Some important functions we will use are:

- `stats.levene(x, y)` performs the equal variance test, with $H_0$: the two populations have the same variance, against $H_1$: they have different variances.
- `stats.ttest_ind(x, y, equal_var, alternative)` performs a two-sample $t$-test for independent samples, testing $\mu_X - \mu_Y$. Use `equal_var=True` if the equal variance test has a large $p$-value, and `equal_var=False` otherwise.
- `stats.ttest_rel(x, y, alternative)` performs a paired (dependent samples) $t$-test. It is the same as a one-sample $t$-test on the differences `x - y`.
- `result.confidence_interval(0.95)` gives the confidence interval from the result of a $t$-test, with its two ends stored in `.low` and `.high`. For `ttest_ind`, it is a confidence interval for $\mu_X - \mu_Y$.
- `stats.probplot(data, plot=plt)` draws a QQ plot to check whether the data are approximately normal.
- `stats.shapiro(data)` performs the Shapiro-Wilk test, with $H_0$: the sample is from a normal distribution, against $H_1$: it is not.

The `alternative` argument is `"two-sided"` (the default), `"greater"` or `"less"`, and it always refers to the first sample minus the second.
