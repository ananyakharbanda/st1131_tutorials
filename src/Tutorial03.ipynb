### Question 3
A university wants to estimate the average number of hours its students spend on part-time jobs per week.

(a) Define the population and a reasonable sampling frame for this study.

Population is the all students enrolled in uni

Reasonable sampling frame is a complete list of all currently enrolled students, such as the university student registry.

(b) Describe how you would draw a simple random sample of 150 students for this study.

Assign every student in the sampling frame a number, then use a random number generator to select 150 distinct students, so every group of 150 students has an equal chance of being selected.

(c) Suppose the university can only access email addresses for students who opted in a mailing list. Explain why this mailing list may not provide an appropriate sampling frame.

The mailing list may cause sampling bias because it excludes students who did not opt in. If students who opt in differ from those who do not in terms of part-time work, the sample may not represent the whole university population.
