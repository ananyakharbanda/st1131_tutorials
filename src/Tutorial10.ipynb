Some important functions we will use are:

* `smf.ols("y ~ x", data=df).fit()` fits a linear regression model by OLS, and `model.summary()` prints the full regression output.
* `model.params`, `model.tvalues` and `model.pvalues` give the coefficient estimates, $t$ values and $p$-values, and `model.fvalue` and `model.f_pvalue` give the $F$-statistic and its $p$-value.
* `model.conf_int(alpha=0.05)` gives the $100(1-\alpha)\%$ confidence intervals for the coefficients.
* `model.predict(new_df)` gives the point estimate of the mean response, and `model.get_prediction(new_df).summary_frame(alpha=0.05)` gives its confidence interval, where `new_df` is a DataFrame with the same column names as in the formula.
* `model.get_influence()` gives the standardised residuals in `.resid_studentized_internal` and the Cook's distances in `.cooks_distance[0]`.
* `model.rsquared` and `model.rsquared_adj` give $R^2$ and the adjusted $R^2$.
* In a formula, `C(var)` treats `var` as a categorical variable, and `I(x**2)` adds the higher-order term $x^2$.
