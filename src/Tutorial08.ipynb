Some important functions we will use are:

- `stats.ttest_1samp(a,popmean,alternative)` gives the $t$-statistic of a 1 sample $t$-test, where `popmean` is the expected value in the null hypothesis $H_0$, and `alternative` defines the alternative hypothesis $H_1$.
- `stats.norm.cdf()` gives $P(Z \le z)$ for $Z \sim N(0,1)$.
- `stats.norm.sf()` gives $P(Z > z)$ for $Z \sim N(0,1)$.
