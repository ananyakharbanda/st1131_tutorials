Some important functions we will use are:

- `comb(n, x)` from `math` gives the binomial coefficient $C^n_x$.
- `stats.binom.pmf(x, n, p)` gives $P(X = x)$ and `stats.binom.cdf(x, n, p)` gives $P(X \le x)$ for $X \sim Bin(n, p)$.
- `stats.norm.cdf(x, loc, scale)` gives $P(X \le x)$, `stats.norm.ppf(q, loc, scale)` gives the $q$-th quantile, and `stats.norm.pdf(x, loc, scale)` gives the density, for $X \sim N(\mu, \sigma^2)$.

For `stats.norm`, `loc` is the mean $\mu$ and `scale` is the **standard deviation** $\sigma$, not the variance $\sigma^2$.

If you imported `binom` and `norm` instead of the entire `stats`, you do not need to write `stats.` at the front of the functions.
