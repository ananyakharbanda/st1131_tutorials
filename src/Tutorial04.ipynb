### Question 1

Two fair six-sided dice are rolled once. Let the outcome of interest be the sum of the numbers shown on the two dice.

(a) Write down the sample space \($S$\) for the sum.

(b) Consider the following events:

- $A$: the sum is an even number.
- $B$: the sum is greater than 8.

Write down the outcomes belonging to each of the following events:

$$
A,\quad B,\quad A \cap B,\quad A \cup B,\quad A^c.
$$

(c) Are events $A$ and $B$ mutually exclusive? Briefly explain your answer.

(d) A student argues:

> "There are 11 possible sums in $S$, so the probability of obtaining a sum of 7 is $\frac{1}{11}$."

Is the student's reasoning correct? If not, find the actual probability of obtaining a sum of 7 and briefly explain your answer.

### Question 5

A screening test is used to detect a rare genetic condition in a population. The condition has a prevalence of 1%. The test has a sensitivity of 96% and a specificity of 95%.

Let:

- $D$: a randomly selected person has the condition.
- $+$: the person receives a positive test result.

(a) Express the prevalence, sensitivity, and specificity using probability notation.

(b) Find the probability of a false positive, $P(+ \mid D^c)$.

(c) Find the probability that a randomly selected person has the condition but receives a negative test result.

(d) Using the Law of Total Probability, find the probability that a randomly selected person receives a positive test result, $P(+)$.

(e) Given that a person receives a positive test result, use Bayes' Theorem to find the probability that the person actually has the condition, $P(D \mid +)$.

(f) A student states:

> "Since the sensitivity of the test is 96%, a person who tests positive has a 96% probability of having the condition."

Is the student's statement correct? Briefly explain your answer.

(g) Suppose the prevalence of the condition decreases, while the sensitivity and specificity of the test remain unchanged. Would $P(D \mid +)$ increase, decrease, or remain unchanged? Briefly explain.
