```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Everything works!")
```

```python
bp = np.array([96, 119, 119, 108, 126, 128, 110, 105, 94])

sorted_bp = np.sort(bp)

print("Sorted data:", sorted_bp)
print("Median:", np.median(bp))
```

The sorted data is 94, 96, 105, 108, 110, 119, 119, 126, 128.
Since there are 9 observations, the median is the 5th observation, which is 110.
The median obtained manually and using Python are the same.

```python
mean_bp = np.mean(bp)
variance_bp = np.var(bp)
sd_bp = np.std(bp)
sample_variance = np.var(bp, ddof=1)

print("Mean:", round(mean_bp, 2))
print("Variance:", round(variance_bp, 2))
print("Standard deviation:", round(sd_bp, 2))
print("Sample variance: ", round(sample_variance, 2))
```

```python
new_bp = bp - 10

new_mean = np.mean(new_bp)
new_variance = np.var(new_bp)

print("New data:", new_bp)
print("New mean:", round(new_mean, 2))
print("New variance:", round(new_variance, 2))
```

The new mean is 101.67 mm Hg.
The new variance is 133.11 (mm Hg)^2.
The mean decreases by 10 because every observation is reduced by 10.
The variance does not change because subtracting the same constant from every observation does not affect the spread of the data.

```python
plt.boxplot(new_bp)

plt.ylabel("Corrected diastolic blood pressure (mm Hg)")
plt.title("Boxplot of Corrected Diastolic Blood Pressure")

plt.show()
```

```python
plt.hist(
    new_bp,
    weights=np.ones(len(new_bp)) / len(new_bp)
)

plt.xlabel("Corrected diastolic blood pressure (mm Hg)")
plt.ylabel("Probability")

plt.show()
```

```python
marks = pd.read_csv("final_marks.csv")
marks.columns.values[1] = "mark"
marks.head()
```

```python
n = marks["mark"].count()

mean_mark = marks["mark"].mean()
median_mark = marks["mark"].median()
variance_mark = marks["mark"].var()

q1 = marks["mark"].quantile(0.25)
q3 = marks["mark"].quantile(0.75)
iqr = q3 - q1

mark_range = marks["mark"].max() - marks["mark"].min()

print("Number of students:", n)
print("Mean:", mean_mark)
print("Median:", median_mark)
print("Variance:", variance_mark)
print("IQR:", iqr)
print("Range:", mark_range)
```

```python
plt.boxplot(marks["mark"])

plt.ylabel("Final mark")
plt.title("Boxplot of Final Marks")

plt.show()
```

```python
q1 = marks["mark"].quantile(0.25)
q3 = marks["mark"].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

outliers = marks[
    (marks["mark"] < lower_fence) |
    (marks["mark"] > upper_fence)
]

outliers
```

Yes, the sample has outliers. The outlying marks are 6 and 7. There are three outlier observations: marks 7, 7 and 6 (corresponding to students 31, 44 and 79).

```python
plt.hist(
    marks["mark"],
    weights=np.ones(len(marks)) / len(marks),
    bins=20,
    density=True,
    edgecolor="black"
)

plt.xlabel("Final mark")
plt.ylabel("Probability")

plt.show()
```

```python
plt.hist(
    marks["mark"],
    bins=5,
    weights=np.ones(len(marks)) / len(marks)
)

plt.xlabel("Final mark")
plt.ylabel("Probability")

plt.show()
```

```python
plt.hist(
    marks["mark"],
    bins=20,
    weights=np.ones(len(marks)) / len(marks)
)

plt.xlabel("Final mark")
plt.ylabel("Probability")

plt.show()
```

By default, the histogram has 10 bins.

Changing the bins argument changes the number of intervals used to display the distribution.
