### Question 2

A university surveyed **200 students** to investigate whether **place of residence** is associated with **preferred mode of transport to campus**. The results are shown below.

| Residence | Public Transport | Walk / Cycle | Private Transport | Total |
|---|:---:|:---:|:---:|:---:|
| On campus | 20 | 55 | 5 | 80 |
| Off campus | 75 | 15 | 30 | 120 |
| **Total** | **95** | **70** | **35** | **200** |

(a) Calculate the percentage distribution of preferred mode of transport separately for students living **on campus** and **off campus**.

* on campus:

public transport = 25%

walk/cycle = 68.75%

private transport = 6.25%

* off campus:

public transport = 62.5%

walk/cycle = 12.5%

private transport = 25%


(b) Construct an appropriate plot to compare the preferred modes of transport between the two groups.

both values are categorical, hence side-by-side (grouped) bar plot

x-axis: Preferred mode of transport

y-axis: Percentage of students

For each transport mode, have two bars: On campus and Off campus

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the table
transport = pd.DataFrame({
    "Public Transport": [20, 75],
    "Walk / Cycle": [55, 15],
    "Private Transport": [5, 30]
}, index=["On campus", "Off campus"])

# Convert counts to percentages within each residence group
transport_pct = transport.div(transport.sum(axis=1), axis=0) * 100

# Transpose so transport mode is on the x-axis
transport_pct.T.plot(kind="bar")

plt.xlabel("Preferred Mode of Transport")
plt.ylabel("Percentage of Students")
plt.title("Preferred Mode of Transport by Place of Residence")
plt.xticks(rotation=0)
plt.legend(title="Residence")

plt.show()
```

(c) Based on your calculations and graphical display, does there appear to be an association between **place of residence** and **preferred mode of transport**? Briefly explain.

Yes, there appears to be an association between place of residence and preferred mode of transport.

The percentage distributions are quite different for the two groups. On-campus students mostly prefer walking/cycling (68.75%), while off-campus students mostly prefer public transport (62.5%). Private transport is also more common among off-campus students (25%) than on-campus students (6.25%).

So, preferred mode of transport seems to depend on where the student lives.


(d) Calculate the difference in proportions and the relative risk of preferring public transport, comparing on-campus and off-campus students. What do these values suggest about the strength of the association?

The difference in proportions is \(0.25-0.625=-0.375\), so on-campus students are 37.5 percentage points less likely to prefer public transport. The relative risk is \(0.25/0.625=0.40\), meaning on-campus students are 0.4 times as likely to prefer public transport as off-campus students. These values indicate a fairly strong association between place of residence and preference for public transport.
