```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/student_survey.csv")
df.head()
df.shape
df.columns
df.dtypes
```

```python
df_clean = df.dropna()
df_clean = df_clean[df_clean["Study_Hours_Per_Week"] > 5]


df_clean["High_GPA"] = np.where(
    df_clean["GPA"] > 4.0,
    "High",
    "No"         
)

df_clean = df_clean.reset_index(drop=True)

df_clean.head()          # first 5 rows
df_clean.tail()          # last 5 rows
print("number of rows")
df_clean.shape           # (number of rows, number of columns)
# df_clean.columns         # column names
df_clean.index           # row index
# df_clean.dtypes          # datatype of each column
# df_clean.info()          # structure + missing-value overview
# df_clean.describe()      # numerical summaries

df_clean.to_csv("../data/new_student.csv", index=True)
```

```python
def descriptive_stats(data, column):
    """
    Return the main descriptive statistics for one quantitative column.

    Use when you want:
    count, mean, median, variance, standard deviation,
    min, Q1, Q3, IQR, max, and range.

    Example:
        descriptive_stats(df, "score")
    """
    s = data[column].dropna()

    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)

    return pd.Series({
        "count": s.count(),
        "mean": s.mean(),
        "median": s.median(),
        "variance": s.var(),
        "std": s.std(),
        "min": s.min(),
        "Q1": q1,
        "Q3": q3,
        "IQR": q3 - q1,
        "max": s.max(),
        "range": s.max() - s.min()
    })

descriptive_stats(df_clean, "GPA")
```
